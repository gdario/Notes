```python
import re

s = 'foo123bar'
re.search('123', s)
```

Match objects like the ones above can be used in boolean expressions.

```python
if re.search('123', s):
    print('Found')

if not re.search('789', s):
    print('Not found')
```

Careful with the use of the `?` metacharacter. It does not mean "either zero or one characters" but rather zero or 1 repetitions of the character immediately before. The first example below finds only the `3` because this matches a `3` preceded by zero repetitions of `1`.

```python
print(re.search('1?3', 'foo123bar'))
print(re.search('1?3', 'foo13bar'))
```


## Other uses of `?`

`?` is also used to create:

1. the non-greedy versions of `*, +, ?`.
2. Lookaheads and lookbehinds.
3. Named groups.

## Metacharacters that match a single character

### Character classes

```python
re.search('ba[artz]', 'foobarqux')
```

Remember, character classes match single characters. Any character in the sequence `bar` would be a match, but only one, the first, is found. This happens because `re.search()` scans the sequence from left to right.

```python
re.search('[a-z]', 'FOObar')
```

In order to match an actual hyphen character in a character class you can do any of the following three things.

1. Place it as the first character (`[-abc]`).
2. Place it as the last character (`[abc-]`).
3. Escape it with a backslash. (`[ab\-c]`)

In order to match a literal `]` in a character class you can:

1. Place it as the first character (`[]]`).
2. Escape it with a backslash (`[ab\]cd]`).

### dot

A single dot matches any single character *except a newline*.

### `\w` and `\W`.

- `\w` matches any alphanumeric *word character*, i.e., uppercase and lowercase letters, digits, and the underscore. It is a shorthand for `[a-zA-Z0-9_]`.
- `\W` is the complement of `\w`, and it is equivalent to `[^a-zA-Z0-9_]`

```python
print(re.search('\w', '#(.a$@&)'))
print(re.search('\W', 'a_1*3Qb'))
```

### `\d` and `\D`

- `\d` matches any character that is a decimal digit. Equivalent to `[0-9]`.
- `\D` matches any character that is *not* a decimal digit. Equivalent to `[^0-9]`.

### `\s` and `\S`

- `\s` matches any whitespace character.
- `\S` matches any character that is *not* a whitespace.

Note, newlines are considered whitespaces.

**Important**: these metacharacters can appear inside brackets, as a character class. The character class `[\d\w\s]` will match the leftmost occurrence of either a decimal digit, a word character, or a whitespace.

### Escaping metacharacters

An expression like `re.search('\\', r'foo\bar')` fails because the double backslash is interpreted first by the Python interpreter and then by the regex parser. The former reads `\\`, interprets it as an escaped `\`, and returns a single `\`, which is incorrectly interpreted by the parser. You need to escape the backslash twice.

```python
print(re.search('\\\\',r'foo\bar'))
```

A better way, is to use a *raw string* to escape the backslash. This prevents the Python interpreter from parsing it returning `\`.

```python
print(re.search(r'\\', r'foo\bar'))
```

## Anchors

### `^` and `$`

`^` and `$` have two corresponding metacharacters used in multiline mode. They are, respectively, `\A` and `\Z`.

### `\b`

This metacharacters can be used to match the beginning or the end of a word. Note that the beginning/end of a word does not necessarily depend on the presence of a whitespace before/after it. This is because

> A word consists of a sequence of alphanumeric characters or underscores (`[a-zA-Z0-9_]`).

```python
print(re.search(r'\bbar', 'foo.bar'))
print(re.search(r'foo\b', 'foo.bar'))
```

```python
print(re.search(r'\bbar\b', 'foo(bar)baz'))
```
